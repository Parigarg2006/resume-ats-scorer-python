```python
import re
from collections import Counter
import math

RESUME_TEXT = """
Rahul Kumar | Python Developer | rahul@email.com
Skills: Python, Machine Learning, TensorFlow, SQL, Django, REST API, Git, Docker
Experience:
- Software Engineer at TechCorp (2021-2023): Developed ML models using scikit-learn and TensorFlow
  Improved model accuracy by 25% using feature engineering and hyperparameter tuning
  Built REST APIs using Django and deployed on AWS EC2
- Junior Developer at StartupXYZ (2019-2021): Built data pipelines using Python and SQL
  Automated reports saving 10 hours/week

Education: B.Tech Computer Science, Delhi University 2019 | CGPA: 8.5/10
Certifications: AWS Solutions Architect, Google ML Certificate
Projects: Fake News Detector (95% accuracy), Stock Price Predictor (LSTM)
"""

JOB_DESCRIPTION = """
We are looking for a Python Developer with experience in machine learning and cloud technologies.
Requirements: Python, TensorFlow, PyTorch, Machine Learning, Deep Learning, SQL, Docker,
AWS, REST API, Git, CI/CD, Agile methodology, Data Science, Pandas, NumPy, scikit-learn
Responsibilities: Build ML models, develop APIs, work with large datasets, deploy on cloud
"""

def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', ' ', text.lower())
    stopwords = {'a','an','the','is','in','it','of','and','to','was','for','on','are','at','with','we','i','you'}
    return [w for w in text.split() if w not in stopwords and len(w) > 2]

def extract_keywords(text, top_n=30):
    words = preprocess(text)
    freq = Counter(words)
    return dict(freq.most_common(top_n))

def calculate_tfidf_score(resume_words, jd_words):
    jd_set = set(jd_words.keys())
    resume_set = set(resume_words.keys())
    matches = jd_set.intersection(resume_set)
    matched_keywords = list(matches)
    missing = jd_set - resume_set
    base_score = len(matches) / len(jd_set) * 100 if jd_set else 0
    return base_score, matched_keywords, list(missing)[:10]

resume_kw = extract_keywords(RESUME_TEXT)
jd_kw = extract_keywords(JOB_DESCRIPTION)
score, matched, missing = calculate_tfidf_score(resume_kw, jd_kw)

print("\n" + "="*55)
print("📄 RESUME ATS ANALYSIS REPORT")
print("="*55)
print(f"\n🎯 ATS Score: {score:.1f}/100")
print(f"\n✅ Matched Keywords ({len(matched)}):")
print("  " + ", ".join(sorted(matched)))
print(f"\n❌ Missing Keywords ({len(missing)}):")
print("  " + ", ".join(sorted(missing)))
print(f"\n📊 Score Breakdown:")
print(f"  Keyword Match Rate: {score:.1f}%")
if score >= 70: print("  ✅ Strong match — likely to pass ATS!")
elif score >= 50: print("  ⚠️  Moderate match — add more keywords")
else: print("  ❌ Weak match — significant keyword gaps")
print(f"\n💡 Suggestions:")
print(f"  Add: {', '.join(list(missing)[:5])}")
```
